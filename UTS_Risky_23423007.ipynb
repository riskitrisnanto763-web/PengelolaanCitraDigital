```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# === 1. Membaca gambar ===
# Ganti path sesuai nama file yang kamu unggah
img = cv2.imread('output-onlinetools.png')

# Konversi dari BGR (default OpenCV) ke RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# === 2. Mengubah ke grayscale ===
gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

# === 3. Membuat citra negatif ===
negatif = 255 - gray

# === 4. Transformasi logaritmik ===
# c = konstanta skala (biasanya 255 / log(1 + Lmax))
c = 255 / np.log(1 + np.max(gray))
log_transform = c * np.log(1 + gray)
log_transform = np.array(log_transform, dtype=np.uint8)

# === 5. Equalisasi histogram ===
hist_eq = cv2.equalizeHist(gray)

# === 6. Menampilkan hasil ===
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title('Citra Asli (RGB)')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(negatif, cmap='gray')
plt.title('Citra Negatif')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(log_transform, cmap='gray')
plt.title('Transformasi Logaritmik')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(hist_eq, cmap='gray')
plt.title('Equalisasi Histogram')
plt.axis('off')

plt.tight_layout()
plt.show()
```
